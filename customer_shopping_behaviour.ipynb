```python
data.drop('Unnamed: 0',axis=1,inplace=True)
```

```python
data.columns=data.columns.str.lower()
data.columns=data.columns.str.replace(' ','_')
data=data.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
#new column age_grp
labels=['youngsters','adult','middle_aged','senior']
data['age_group']=pd.qcut(data['age'],q=4,labels=labels)
#qcut splits the  ages into 4 equal size grps
```

```python
data[['age','age_group']].sample(5)
```

```python
data['frequency_of_purchases'].unique()
```

```python
#new column purchase_frequnecy_days - to convert the purchase freq in text form to numerical by no.of days
frequency_mapping={'Fortnightly':14,
                   'Weekly':7,
                   'Annually':365,
                   'Quarterly':90,
                   'Bi-Weekly':14,
                   'Monthly':30,
                   'Every 3 Months':90}
data['purchase_frequnecy_days']=data['frequency_of_purchases'].map(frequency_mapping)
```

```python
data[['purchase_frequnecy_days','frequency_of_purchases']].head(10)
```

```python
data[['discount_applied','promo_code_used']].head(10)
```

```python
(data['discount_applied']==data['promo_code_used']).all()
#whenever promo code used discount is availed..both columns have same data so can drop one column
```

```python
data.drop('promo_code_used',axis=1,inplace=True)
```
